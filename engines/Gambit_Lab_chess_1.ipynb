## Step 4 — this cell *is* the game

Type your move in the box on the right (normal notation: `e4`, `Nf3`, `O-O` — or coordinates like `e2e4`), then run the cell. Stockfish answers and the board redraws with its move arrowed. **Run this cell once per move.**

```python
your_move = "e4" #@param {type:"string"}

def parse_move(board, text):
    text_clean = text.strip()
    # Try Standard Algebraic Notation (SAN) first
    try:
        return board.parse_san(text_clean)
    except ValueError:
        pass
    # Try Universal Chess Interface (UCI) coordinate notation
    try:
        return board.parse_uci(text_clean.lower())
    except ValueError:
        pass
    return None

try:
    # Safety check: ensure game board is initialized
    if 'board' not in globals():
        print("🔄 Game board was not initialized. Creating a new game now!")
        board = chess.Board()

    if board.is_game_over():
        print("Game over:", board.result(), "— run Step 3 for a new game.")
    else:
        m = parse_move(board, your_move)
        if m is not None and m in board.legal_moves:
            board.push(m)
            clear_output(wait=True)

            # Engine response
            if not board.is_game_over():
                # Ensure engine is active and configured
                try:
                    engine.analyse(chess.Board(), chess.engine.Limit(time=0.01))
                except (NameError, chess.engine.EngineTerminatedError, AttributeError):
                    engine = chess.engine.SimpleEngine.popen_uci(ENGINE_PATH)
                    set_elo = max(1320, min(3190, TARGET_ELO if 'TARGET_ELO' in globals() else 1350))
                    engine.configure({"UCI_LimitStrength": True, "UCI_Elo": set_elo})

                result = engine.play(board, chess.engine.Limit(time=0.1))
                stockfish_san = board.san(result.move)
                board.push(result.move)

                show(board, last_move=result.move)
                print(f"You played: {your_move}")
                print(f"Stockfish played: {stockfish_san}")
            else:
                show(board)
                print("Game Over!")
        else:
            # List available moves safely
            legal_options = ", ".join([board.san(x) for x in board.legal_moves])
            print(f"❌ Illegal or invalid move: '{your_move}'")
            print(f"Please try a valid move. Legal options are: {legal_options}")
except Exception as e:
    print("An unexpected error occurred:", e)
    print("If the engine crashed, please re-run Step 2.")
```

#@title Step 5 (optional) · Whole game in one cell — type moves as prompted
This block lets you play a full game continuously with live interactive prompt inputs.

```python
import time
board = chess.Board()

try:
    while not board.is_game_over():
        clear_output(wait=True)
        show(board)

        # Safely prompt the user for input without breaking on quick clicks
        try:
            user_input = input("Your move (or 'undo', 'resign'): ").strip()
        except KeyboardInterrupt:
            print("\nGame paused. Run this cell again to continue or start over.")
            break

        if user_input.lower() == 'resign':
            print("You resigned. Game over!")
            break
        elif user_input.lower() == 'undo':
            if len(board.move_stack) >= 2:
                board.pop()
                board.pop()
            continue

        try:
            try:
                m = board.parse_san(user_input)
            except ValueError:
                m = chess.Move.from_uci(user_input.lower())

            if m in board.legal_moves:
                board.push(m)
                if not board.is_game_over():
                    result = engine.play(board, chess.engine.Limit(time=0.15))
                    board.push(result.move)
            else:
                print(f"Illegal move: {user_input}. Try again.")
                time.sleep(1.5)
        except Exception as move_err:
            print(f"Invalid move notation or error: {move_err}")
            time.sleep(1.5)

    if board.is_game_over():
        clear_output(wait=True)
        show(board)
        print("Game ended.", board.result())
except Exception as main_err:
    print(f"An unexpected error occurred during the game loop: {main_err}")
```

Run the cell below to watch the 1350 Elo engine play a 60 half-move game against the 2400 Elo engine on a self-updating board, followed by a final evaluation by the coach.

```python
import time
from IPython.display import display, HTML, clear_output
import chess
import chess.svg
import chess.engine

# Initialize the engine instances
weak_engine = chess.engine.SimpleEngine.popen_uci(ENGINE_PATH)
strong_engine = chess.engine.SimpleEngine.popen_uci(ENGINE_PATH)

# Configure their strengths (Strong engine increased to 3000 Elo)
weak_engine.configure({"UCI_LimitStrength": True, "UCI_Elo": 1350})
strong_engine.configure({"UCI_LimitStrength": True, "UCI_Elo": 3000})

battle_board = chess.Board()
max_half_moves = 60
move_count = 0
score_val = 0.0  # Normalized evaluation score for the bar (0.5 is equal)

def render_game_layout(board, score_percent, eval_text, last_move=None):
    """Generates a side-by-side HTML display of an evaluation bar and the chessboard."""
    board_svg = chess.svg.board(board, size=400, lastmove=last_move)

    # Ensure the bar percentage is within valid bounds [0, 100]
    clamped_pct = max(0, min(100, score_percent))

    # Build side-by-side layout with flexbox
    html_output = f"""
    <div style="display: flex; align-items: center; gap: 20px; font-family: sans-serif;">
        <!-- Evaluation Bar Container (Black on bottom, White on top) -->
        <div style="position: relative; width: 30px; height: 400px; border: 2px solid #333; border-radius: 4px; overflow: hidden; background-color: #222;">
            <div style="position: absolute; bottom: 0; left: 0; width: 100%; height: {clamped_pct}%; background-color: #f0f0f0; transition: height 0.3s ease;"></div>
            <div style="position: absolute; top: 50%; left: 0; width: 100%; height: 2px; background-color: #ff3b30; z-index: 2; opacity: 0.8;"></div>
        </div>
        <!-- Chessboard Column -->
        <div>
            {board_svg}
        </div>
    </div>
    <div style="margin-top: 10px; font-family: sans-serif; font-size: 14px;">
        <strong>Evaluation:</strong> {eval_text}
    </div>
    """
    display(HTML(html_output))

try:
    # Play up to 60 half-moves
    while not battle_board.is_game_over() and move_count < max_half_moves:
        if battle_board.turn == chess.WHITE:
            active_engine = weak_engine
            role_name = "⚪ White (Weak - 1350 Elo)"
        else:
            active_engine = strong_engine
            role_name = "⚫ Black (Strong - 3000 Elo)"

        # Engine computes move
        result = active_engine.play(battle_board, chess.engine.Limit(time=0.05))
        san_move = battle_board.san(result.move)

        # Execute move
        battle_board.push(result.move)
        move_count += 1

        # Analyze current position for the evaluation bar (using the stronger engine)
        analysis = strong_engine.analyse(battle_board, chess.engine.Limit(time=0.02))
        score = analysis["score"].white()

        # Normalize score into a percentage for the evaluation bar
        if score.is_mate():
            mate_moves = score.mate()
            eval_text = f"Mated in {-mate_moves if mate_moves < 0 else mate_moves} moves"
            score_percent = 100 if mate_moves > 0 else 0
        else:
            cp = score.score() / 100.0  # Centipawns to pawns
            eval_text = f"{'+' if cp >= 0 else ''}{cp:.2f}"
            # Map pawn advantage [-5, +5] into [0, 100] percentage
            score_percent = ((cp + 5) / 10.0) * 100

        # Update display smoothly
        clear_output(wait=True)
        render_game_layout(battle_board, score_percent, eval_text, last_move=result.move)
        print(f"🎬 Match Progress: [{move_count}/{max_half_moves} half-moves played]")
        print(f"👉 {role_name} played: {san_move}")

        # Configured 2.0-second delay as requested
        time.sleep(2.0)

    print("\n🏁 Match Finished or reached move limit!")
    print("Final Game Status:", battle_board.result() if battle_board.is_game_over() else "Adjourned (completed 60 half-moves)")

    # Post-Game Coach Analysis of the final board state
    print("\n--- 🧠 Coach's Post-Game Position Evaluation ---")
    info = strong_engine.analyse(battle_board, chess.engine.Limit(time=1.0))
    score = info["score"].white()
    print(f"Position Evaluation (from White's perspective): {score}")

    hint = info.get("pv", [None])[0]
    if hint:
        print(f"Best move from this final position: {battle_board.san(hint)}")

finally:
    # Always release engine resources
    weak_engine.quit()
    strong_engine.quit()
```
